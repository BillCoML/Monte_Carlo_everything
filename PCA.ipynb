**To understand most of it, you should know**

1. Linear Transformation
2. Langrange Multipliers
3. Eigenvectors and Eigenvalues

**1. The goal**

Notice that, in our dataset, we are using the standard basis system.

$$
e_1 = \begin{pmatrix}
0 \\ 1
\end{pmatrix}, 

e_2 = \begin{pmatrix}
1 \\ 0
\end{pmatrix}
$$

In higher dimension, there will be more of those basis and each as well will have more items.

**Our goal** is to find some vectors in space that, when we project our data onto it, the variance explained by each of those axes will be optimized.

Each of those axes, should encode information using all the features in the current basis system.

**E.g:** There are 3 axis; height, weight and IQ. We want just 2 combined features. So, our goal is to have the 1st new axis that encodes data from both **height, weight and IQ** in a way that the variance formed by that is pretty high.

**2. Change of Basis, Covariance matrix**

We want a transformation to another basis system (change of basis). 

In our basis, a sample $x_1$ in 2D can be represent as:
$$
x_1 = x_{11}\begin{pmatrix}
0 \\ 1
\end{pmatrix} +

x_{12} \begin{pmatrix}
1 \\ 0
\end{pmatrix}
$$

where $x_{11}, x_{12}$ are recorded values for each of the features for this sample.

** Note that, data is usually much larger than just 2D, that's when PCA can shine.

Because we stated that after the **transformation**, we want it to have the most variance. 

And instead of using the old basis, we want to find the new one(s) that maximize the **transformed variance**. Thus, even using less axis still somehow reserve enough information.

Let the new basis be 

$$
v_1, v_2, ..., v_n
$$

Where $n$ is number of current features. And they all form the transformation matrix.

Then the transformation onto each of the new axis $i$ will be:

$$
v_i \cdot X
$$

Here I use big $X$ instead of $x$ because I am referring to the whole data matrix.

Then the variance of the new axis $i$ is 
$$
Var(v_i \cdot X) = v_i ^ T \cdot \Sigma \cdot v_i
$$

Here, $\Sigma$ is the Covariance matrix of data $X$ (how each of the features is correlated with the others).

**3. Constrain and Maximize**

We need to impose a constraint on this, which is, the transform vector $v_i$ needs to be of unit length:

$$
v_i^T \cdot v_i = 1, \text{all }i
$$

We will see why this constraint is used (even though you will also see it doesn't affect the optimization step).

Using Langrage Multipliers:

$$
\nabla (v_i ^ T \cdot \Sigma \cdot v_i) = \lambda\nabla (v_i^T \cdot v_i) \\

\rightarrow \Sigma v_i = \lambda v_i
$$

Well, this is exactly the eigenvector and eigenvalue identity.

Now, let's see why $v_i^T\cdot v_i = 1$ is used.

+ Since $\Sigma$, the covariance matrix, is symmetric. Then, all eigenvectors are orthogonal.
+ By adding the unit vector condition. We will achieve:

$$
V \cdot V^T = I
$$

$$
\text{Where, } V = 
\begin{bmatrix}
v_{11} & v_{21} & v_{31} \\
v_{12} & v_{22} & v_{32} \\
v_{13} & v_{23} & v_{33}
\end{bmatrix}
$$

So, we conclude $V^T = V^{-1}$

**4. New variance, and reduce dimensionality**

We have found:
+ $\Sigma v_i = \lambda v_i$
+ $V^T = V^{-1}$

So, agree with me that 

$$
v_i^T \Sigma v_i = \lambda
$$

or, if you prefer to see the big picture:

$$
V^T\cdot \Sigma \cdot V = \lambda I
$$

The left hand side is exactly $Var(v_i \cdot X)$, it means each eigenvalue corresponding to eigenvector $i^{th}$ is simply the new variance. And the largest eigenvalue is the largest variance.

Now, we have a list of eigenvalues from top to bottom, by choosing those that really explain much of the variance, we can effectively reduce the dimension of our data.
