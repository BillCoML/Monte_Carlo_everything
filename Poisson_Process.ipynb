### Example 1

An office receives phone calls, on <i>average 6 calls per hour</i>. Given that in the first 1.5 hours, there was **exactly 5 calls**. What is the probability that **those 5 calls both occured in the first 1 hour** ?

```python
mu = 6
```

<u>Analytical</u>



We know that # phone calls follows a Poisson distribution. Each phone call is independent of each other, and thus number of phone calls per $\triangle t$ is also independent.

Let $N(1)$ model # calls during (0, 1]

$$N(1) \sim Pois(\lambda\cdot1)$$

$N(1.5)-N(1)$ models # calls during (1, 1.5]

$$N(1.5) - N(1) \sim Pois(\lambda\cdot0.5)$$

Of course $N(1.5)$ is during (0, 1.5]
$$N(1.5) \sim Pois(\lambda\cdot1.5)$$

This is our strategy
$$P(N(1)=5 | N(1.5)=5) = \frac{P(N(1)=5 \cap N(1.5) - N(1) = 0)}{P(N(1.5) = 5)}$$

```python
p = poisson.pmf(5, mu=mu*1) * poisson.pmf(0, mu=mu*0.5) / poisson.pmf(5, mu=mu*1.5)

print("Our exact probability:", p)
```

<u>Numerical</u>

We will see how many of those that, in the period (0, 1] there are exactly 5. And during (1, 1.5] exactly 0.

$$
P(N(1)=5 | N(1.5)=5) = \frac{P(N(1)=5 \cap N(1.5) - N(1) = 0)}{P(N(1.5) = 5)}
$$

$P(N(1)=5 \cap N(1.5) - N(1) = 0)$

```python
n = 1000000

period1 = poisson.rvs(mu = mu * 1, size = n)

period2 = poisson.rvs(mu = mu * 0.5, size = n)
```

```python
mask1 = (period1 == 5) & (period2 == 0)
```

$P(N(1.5) = 5)$

```python
all_time = poisson.rvs(mu = mu* 1.5, size = n)
```

```python
mask2 = (all_time == 5)
```

$P(N(1)=5 | N(1.5)=5)$

```python
#The answer (very close)

p = np.mean(mask1) / np.mean(mask2)

print(p)
```
