**Negative Binomial-Poisson**

Let X be a random variable that follows Negative Binomial, it models # failures until rth success.

$$
X \sim NB(r, p)
$$

Under 3 main conditions, $X \to Poisson(\lambda)$
1. $r \to \infty$
2. $p \to 1$
3. $r(1-p) \to \lambda$

Let's go through an example, it's very common that mosquitoes get into a specific house. As recorded, 99% of the days, except for a few extremely cold periods, that house will encounter mosquitoes. 

Thus, we can say that the probability of that house sees mosquitoes every day is:

$$p \approx 0.99$$

Let's say the owner is interested, and he writes down whether mosquitoes fly in every single day. 

Then, how many days will go by **without mosquitoes**, until the 1000th day of having mosquitoes.

The, let X denotes number of days without mosquitoes.
$$X \sim NB(r=1000, p=0.999)$$

Since both r and p kinda approach its limit, then $\lambda=r(1-p) = 1$

```python
from scipy.stats import nbinom, poisson
```

We can try find, what is the probability that there is only 2 days without mosquitoes. The 2 are very close to each other

```python
#Negative Binomial (Exact)
print(nbinom.pmf(2, 1000, 0.999))

#Poisson (Approx)
print(poisson.pmf(2, 1))
```

**Gamma-Poisson**

This pair shows up in many cases, one important (very handful) case is in Bayesian Statistics.

+ <u>Case: Cumulative function</u>

If we want to calculate Cumulative probability of a random variable Z in the form:

$$
F_Z(z = x) \\
Z \sim Gamma(\alpha, 1)
$$

We can always use its identity of a Poisson:

$$
Y \sim Poisson(\lambda = x)
$$

The identity is as follow:

$$
\int_x^\infty \frac{1}{\Gamma(\alpha)} z^{\alpha - 1} e^{-z} \, dz = \sum_{y=0}^{\alpha-1} \frac{e^{-x}x^y}{y!}
$$

Remember a useful way of calculating CDF of Poisson, where:
$$
P(X = x + 1) = \frac{\lambda}{x + 1}P(X = x), \\ 
\text{Where, } P(X = 0) = e^{-\lambda}
$$
