**Set theory**

Back to the definition of sets, random variables are built upon some experiment with defined outcomes called the sample space $\Omega$.

E.g: Tossing 2 dice, then $\Omega = {11, 12, 13, 14, ..., 66}$. there are 36 possible outcomes. 

Then we would assign probability to each of the outcome, so if we assume this is a fair dice, it means each score has equal probability of $p=\frac{1}{6}$. Thus each outcome has $p=\frac{1}{36}$.

**Sigma Algebra**

Simply think of it as all possible subsets. If you have 3 outcomes A,B,C, then we have $2^3 = 8$ subsets, including $\emptyset$, the empty set. We call them **the events**. 

We said ealier that each outcome has an assigned probability, like score of $1, 3$ has probability $\frac{1}{6}\cdot\frac{1}{6}=\frac{1}{36}$.

Then each event (set of outcomes) has probability of sum of probability of all outcomes in that set.

E.g: $A_1 = \{(1,1), (3,4)\}$ has probability $\frac{1}{36} + \frac{1}{36} = \frac{2}{36}$

**Random Variable**

Random Variable defines a mapping from some event in Sigma Algebra to a realization of $X$. 

Only those events with all outcomes satisfy the Random Variable are used, and no event is used more than once when mapping. Thus, we term it **Mutual Exclusiveness**, or sometimes **Disjoint**. Intersection of 2 disjoint sets automatically results in an empty set, they share nothing in common.


So, if we define $X$ as sum of scores of 2 dice. Then its realization (some says Sample Space of $X$) is

$$S_x = \{2,3,4,...,12\}$$

And for $X = 3$, we can trace back to the Sigma Algebra, taking an event in which all elements sum to 3.

E.g: $X=3$, then $X^{-1}(3) = \{(1,2),(2,1)\}$. This event has probability $p=\frac{2}{36}$, thus is also the value of $P(X=3)$.

**Conditioning**

We can condition $Y$ on $X$ if they are defined from the same Sample Space $\Omega$. This will make a lot of sense in a second.

Again, our conditional probability has form

$$
f_X(x | y) = \frac{f_{X, Y}(x, y)}{f_Y(y)}
$$

where,

+ $f_X(x|y)$: the probability of $X$, given information of $Y$, also the value we are looking for.
+ $f_{X,Y}(x, y)$: the joint probability, literally $P(X, Y)$.
+ $f_Y(y)$: marginal probability of $Y$, or simply probability of $Y$.

Think of it like this, if $Y=y$ is a realization that corresponds to a set of outcomes. Then $X,Y = X \cap Y$ has to be a child set of $Y=y$.

E.g:

$X$ defines sum of 2 dice, $Y$ defines max of 2 dice.

We want to calculate $P(X = 5|Y = 4)$.

Now, 

+ $Y = 4$, which is the set $\{(1,4),(2,4),(3,4),(4,1),(4,2),(4,3),(4,4)\}$, probability is $p = 7 \cdot \frac{1}{36}$

+ $X = 5$ belongs to $\{(1,4),(2,3),(3,2),(4,1)\}$

+ Thus, $X \cap Y$ is $\{(1,4),(4,1)\}$, which has probability $p=2\cdot\frac{1}{36}$.

So $P(X = 5|Y=4) = \frac{2/36}{7/36} = \frac{2}{7}$.


Understood? So easy. Now think, if they don't have the same sample space, it makes no sense to find conditional probability.

**Independence**

If $P(X|Y) = P(X)$, it means knowing Y doesn't help reduce $P(X)$. So X and Y are independent. Otherwise, they are dependent.
